```python
# Removing unwanted columns 
df = df.drop(['rating_count','lic_no', 'link'], axis = 1 )
df.head()
```

```python
# Removing duplicate values 
df.drop_duplicates(inplace = True )
df.shape
```

```python
df["rating"].unique()
```

```python
# Replace '--' with NaN
df['rating'] = df['rating'].replace('--', np.nan)

# Convert the column to numeric
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
```

```python
df["rating"].fillna(df["rating"].mean(),inplace = True )
```

```python
df.dropna(inplace = True)
df.head()
```

```python
df['cost'] = (
    df['cost']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
    .astype(float)
)
```

```python
#Cleaning cuisine
cuisine = df["cuisine"].value_counts(ascending = False)
cuisine
```

```python
cuisine_lessthan1000 = cuisine[cuisine<1000]
cuisine_lessthan1000
```

```python
def handle_cuisine(value):
    if (value in cuisine_lessthan1000):
        return "other"
    else:
        return value
df["cuisine"] = df["cuisine"].apply(handle_cuisine)
df["cuisine"].value_counts()
```

```python
df.head()
```

```python
address = df["address"].value_counts(ascending = False)
address_lessthan300 = address[address<300]

def handle_address(value):
    if(value in address_lessthan300):
        return "others"
    else:
        return value

df["address"] = df["address"].apply(handle_address)
df["address"].value_counts()
```

```python
cuisine = df["cuisine"].value_counts(ascending = False)

cuisine_lessthan100 = cuisine[cuisine<100]

def handle_cuisine(value):
    if(value in cuisine_lessthan100):
        return "others"
    else:
        return value
df["cuisine"] = df["cuisine"].apply(handle_cuisine)
df["cuisine"].value_counts()
```
